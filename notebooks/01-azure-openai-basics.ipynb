# 01 — Create Azure OpenAI embeddings
This lab calls the deployed `text-embedding-3-large` model directly. No database is used until the final exercise.

```python
from os import getenv
from dotenv import load_dotenv
from openai import OpenAI
load_dotenv('../.env')
endpoint = getenv('AZURE_OPENAI_EMBEDDING_ENDPOINT') or getenv('AZURE_OPENAI_ENDPOINT').replace('.openai.azure.com', '.cognitiveservices.azure.com')
deployment = getenv('AZURE_OPENAI_EMBEDDING_DEPLOYMENT')
client = OpenAI(api_key=getenv('AZURE_OPENAI_API_KEY'), base_url=endpoint.rstrip('/') + '/openai/v1/')
print({'endpoint_configured': bool(endpoint), 'deployment': deployment})
```

```python
phrases = ['Model Context Protocol connects models to tools.', 'GitHub Copilot helps developers write code.', 'A mango is a tropical fruit.']
response = client.embeddings.create(model=deployment, input=phrases)
vectors = [item.embedding for item in response.data]
print('Embedding dimensions:', len(vectors[0]))
print('Tokens used:', response.usage.total_tokens)
```

```python
import math
def cosine(a, b): return sum(x*y for x,y in zip(a,b)) / math.sqrt(sum(x*x for x in a)*sum(y*y for y in b))
for i in range(1, len(phrases)):
    print(phrases[0], 'vs', phrases[i], round(cosine(vectors[0], vectors[i]), 3))
```
