# 10 — Clone and run the Student MCP server
This notebook is the participant onboarding guide. The GitHub repository does **not** include `.venv` or any executable files. Each participant creates their own virtual environment and MCP executable.

## 1. Clone the public repository
Run in PowerShell, not in a notebook cell:

```powershell
git clone https://github.com/DeccansoftAITeam/BITC-Graph-Chatbot.git
cd BITC-Graph-Chatbot
```

## 2. Create the virtual environment and install the project

```powershell
py -m venv .venv
.\.venv\Scripts\Activate.ps1
python -m pip install --upgrade pip
python -m pip install -e .
```

This creates `course-knowledge-student-mcp.exe` inside `.venv\Scripts`.

## 3. Configure workshop credentials

```powershell
Copy-Item .env.workshop.example .env
```

Fill the supplied participant values. Never commit `.env`. Students should receive a read-only workshop PostgreSQL account and restricted Azure keys, not instructor credentials.

````python
from pathlib import Path
mcp_exe = Path('.venv/Scripts/course-knowledge-student-mcp.exe')
print({'mcp_executable_exists': mcp_exe.exists(), 'env_exists': Path('.env').exists()})
````

## 4. Configure Claude Desktop or Antigravity
Replace `C:\\path\\to\\BITC-Graph-Chatbot` with the participant's actual clone location:

```json
{
  "mcpServers": {
    "bestitcourses-learning": {
      "command": "C:\\path\\to\\BITC-Graph-Chatbot\\.venv\\Scripts\\course-knowledge-student-mcp.exe",
      "cwd": "C:\\path\\to\\BITC-Graph-Chatbot"
    }
  }
}
```

Restart the MCP client after saving its configuration.

Ask the client: **“Using BestITCourses, what is Model Context Protocol?”**
