# 03 — Keyword + vector + RRF
Compare retrieval legs before they are fused. RRF rewards chunks that rank well in multiple independent lists.

```python
from course_knowledge.database import connect, keyword_candidates, vector_candidates
from course_knowledge.embeddings import AzureOpenAIEmbedder
from course_knowledge.settings import database_settings, azure_openai_settings
from course_knowledge.hybrid import HybridRetriever
question = 'What is Model Context Protocol?'
with connect(database_settings()) as db:
    embedder = AzureOpenAIEmbedder.from_settings(azure_openai_settings())
    vector = embedder.embed([question])[0]
    print('KEYWORD:', [x['canonical_filename'] for x in keyword_candidates(db, question, 5)])
    print('VECTOR:', [x['canonical_filename'] for x in vector_candidates(db, vector, 5)])
    for hit in HybridRetriever(db, embedder).search(question, 5): print('RRF', hit.matched_legs, hit.candidate.metadata['source'])
```
